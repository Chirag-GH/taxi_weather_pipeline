## Borough ID Table

```python
borough_schema = StructType([
    StructField("borough_id", IntegerType(), False),
    StructField("name", StringType(), False)
])

borough_data = [
    (1, "Manhattan"),
    (2, "Brooklyn"),
    (3, "Queens"),
    (4, "Bronx"),
    (5, "Staten Island"),
    (6, "EWR"),
    (99, "Unknown")
]

silver_borough = spark.createDataFrame(borough_data, schema=borough_schema)
```

## Payment Type Table

```python
payment_schema = StructType([
    StructField("payment_type", IntegerType(), False),
    StructField("type", StringType(), False)
])

payment_data = [
    (0, "Flex Fare trip"),
    (1, "Credit card"),
    (2, "Cash"),
    (3, "No charge"),
    (4, "Dispute"),
    (5, "Unknown"),
    (6, "Voided trip")
]

silver_payment = spark.createDataFrame(payment_data, schema=payment_schema)
```

# Write to Delta Tables

```python
silver_wc.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_wc_dim")

silver_borough.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_borough_dim")

silver_payment.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_payment_dim")

silver_vendor.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_vendor_dim")

silver_rate.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_rate_dim")
```
