# Required Libraries and Functions

```python
from pyspark.sql.functions import current_timestamp, col
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
```

# Hardcoded mapping

## Weather Code Table

```python
wc_schema = StructType([
    StructField("weather_code", IntegerType(), False),
    StructField("description", StringType(), False)
])

wc_data = [
    (0, "Clear sky"),
    (1, "Mainly clear"),
    (2, "Partly cloudy"),
    (3, "Overcast"),
    (45, "Fog"),
    (48, "Depositing rime fog"),
    (51, "Light drizzle"),
    (53, "Moderate drizzle"),
    (55, "Dense drizzle"),
    (56, "Light freezing drizzle"),
    (57, "Dense freezing drizzle"),
    (61, "Slight rain"),
    (63, "Moderate rain"),
    (65, "Heavy rain"),
    (66, "Light freezing rain"),
    (67, "Heavy freezing rain"),
    (71, "Slight snowfall"),
    (73, "Moderate snowfall"),
    (75, "Heavy snowfall"),
    (77, "Snow grains"),
    (80, "Slight rain showers"),
    (81, "Moderate rain showers"),
    (82, "Violent rain showers"),
    (85, "Slight snow showers"),
    (86, "Heavy snow showers"),
    (95, "Thunderstorm"),
    (96, "Thunderstorm with slight hail"),
    (97, "Heavy thunderstorm"),
    (99, "Thunderstorm with heavy hail")
]

silver_wc = spark.createDataFrame(wc_data, schema=wc_schema)
```

## Borough ID Table

```python
borough_schema = StructType([
    StructField("borough_id", IntegerType(), False),
    StructField("name", StringType(), False)
])

borough_data = [
    (1, "Manhattan"),
    (2, "Brooklyn"),
    (3, "Queens"),
    (4, "Bronx"),
    (5, "Staten Island"),
    (6, "EWR"),
    (99, "Unknown")
]

silver_borough = spark.createDataFrame(borough_data, schema=borough_schema)
```

## Payment Type Table

```python
payment_schema = StructType([
    StructField("payment_type", IntegerType(), False),
    StructField("type", StringType(), False)
])

payment_data = [
    (0, "Flex Fare trip"),
    (1, "Credit card"),
    (2, "Cash"),
    (3, "No charge"),
    (4, "Dispute"),
    (5, "Unknown"),
    (6, "Voided trip")
]

silver_payment = spark.createDataFrame(payment_data, schema=payment_schema)
```

## Vendor ID Table

```python
vendor_schema = StructType([
    StructField("VendorID", IntegerType(), False),
    StructField("name", StringType(), False)
])

vendor_data = [
    (1, "Creative Mobile Technologies, LLC"),
    (2, "Curb Mobility, LLC"),
    (6, "Myle Technologies Inc"),
    (7, "Helixx")
]

silver_vendor = spark.createDataFrame(vendor_data, schema=vendor_schema)
```

## Rate Code Table

```python
rate_schema = StructType([
    StructField("RatecodeID", IntegerType(), False),
    StructField("description", StringType(), False)
])

rate_data = [
    (1, "Standard rate"),
    (2, "JFK"),
    (3, "Newark"),
    (4, "Nassau or Westchester"),
    (5, "Negotiated fare"),
    (6, "Group ride"),
    (99, "Unknown")
]

silver_rate = spark.createDataFrame(rate_data, schema=rate_schema)
```

# Write to Delta Tables

```python
silver_wc.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_wc_dim")

silver_borough.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_borough_dim")

silver_payment.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_payment_dim")

silver_vendor.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_vendor_dim")

silver_rate.write.format("delta").mode("overwrite").saveAsTable("workspace.taxi_weather.silver_rate_dim")
```
